### Project Overview:

A bank's credit card department is one of the top adopters of data science. A top focus for the bank has always been acquiring new credit card customers. Giving out credit cards without doing proper research or evaluating applicants' creditworthiness is quite risky. The credit card department has been using a data-driven system for credit assessment called Credit Scoring for many years, and the model is known as an application scorecard. A credit card application's cutoff value is determined using the application scorecard, which also aids in estimating the applicant's level of risk. This decision is made based on strategic priority at a given time
	Customers must fill out a form, either physically or online, to apply for a credit card. The application data is used to evaluate the applicant's creditworthiness. The decision is made using the application data in addition to the Credit Bureau Score, such as the FICO Score in the US or the CIBIL Score in India, and other internal information on the applicants. Additionally, the banks are rapidly taking a lot of outside data into account to enhance the caliber of credit judgements.

```python
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.optimizers import Adam, RMSprop, SGD
from keras.callbacks import EarlyStopping
import optuna
```

```python
def objective(trial):
    # Hyperparameters to optimize
    num_layers = trial.suggest_int('num_layers', 1, 7)
    units = trial.suggest_int('units', 32, 128, step=16)
    activation_function = trial.suggest_categorical('activation_function', ['relu', 'leaky_relu', 'sigmoid', 'tanh'])
    learning_rate = trial.suggest_float('learning_rate', 1e-5, 1e-1, log=True)
    batch_size = trial.suggest_int('batch_size', 32, 128, step=32)
    optimizer = trial.suggest_categorical('optimizer', ['adam', 'sgd', 'rmsprop'])
    epochs = trial.suggest_int('epochs', 10, 100)
    dropout_rate = trial.suggest_float('dropout_rate', 0.1, 0.5)

    # Create model
    model = Sequential()

    # First hidden layer
    model.add(Dense(units=units, activation=activation_function, input_dim=X_train.shape[1]))
    model.add(Dropout(dropout_rate))

    # Add additional layers
    for _ in range(num_layers - 1):
        model.add(Dense(units=units, activation=activation_function))
        model.add(Dropout(dropout_rate))

    # Output layer
    model.add(Dense(1, activation="sigmoid"))

    # Select optimizer
    if optimizer == "adam":
        optimizer_instance = Adam(learning_rate=learning_rate)
    elif optimizer == "sgd":
        optimizer_instance = SGD(learning_rate=learning_rate)
    elif optimizer == "rmsprop":
        optimizer_instance = RMSprop(learning_rate=learning_rate)

    # Compile the model
    model.compile(optimizer=optimizer_instance, loss="binary_crossentropy", metrics=["accuracy"])

    # Train the model
    history = model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, validation_data=(X_test, y_test), verbose=0)

    # Return validation accuracy as the optimization objective
    val_accuracy = history.history['val_accuracy'][-1]
    return val_accuracy

# Create a study object and optimize the objective function
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=50)  # You can set n_trials as per your requirement

# Get the best hyperparameters found by Optuna
best_params = study.best_params
print("Best hyperparameters:", best_params)
```

```python
from optuna.visualization import plot_optimization_history, plot_parallel_coordinate, plot_slice, plot_contour, plot_param_importances
plot_optimization_history(study).show()
```

```python
import optuna.visualization as vis

# Plot the optimization history
vis.plot_optimization_history(study)

# Plot the hyperparameter importance
vis.plot_param_importances(study)
```

```python
# Assuming you have already run Optuna and obtained the best parameters
best_params = study.best_params

# Create and compile the model with the best hyperparameters
model = Sequential()

# First hidden layer
model.add(Dense(units=best_params['units'],
                activation=best_params['activation_function'],
                input_dim=X_train.shape[1]))  # Use the input dimension of your training data
model.add(Dropout(best_params['dropout_rate']))  # Apply dropout

# Add additional hidden layers
for _ in range(best_params['num_layers'] - 1):  # Subtract 1 because we already added one layer above
    model.add(Dense(units=best_params['units'],
                    activation=best_params['activation_function']))
    model.add(Dropout(best_params['dropout_rate']))  # Apply dropout after each hidden layer

# Output layer (binary classification)
model.add(Dense(1, activation='sigmoid'))  # Sigmoid activation for binary classification

# Select optimizer based on the best parameter
if best_params["optimizer"] == "adam":
    optimizer = Adam(learning_rate=best_params["learning_rate"])
elif best_params["optimizer"] == "sgd":
    optimizer = SGD(learning_rate=best_params["learning_rate"])
elif best_params["optimizer"] == "rmsprop":
    optimizer = RMSprop(learning_rate=best_params["learning_rate"])

# Compile the model
model.compile(optimizer=optimizer, loss="binary_crossentropy", metrics=["accuracy"])

# Train the model with the best hyperparameters
history = model.fit(X_train, y_train,
                    epochs=best_params["epochs"],
                    batch_size=best_params["batch_size"],
                    validation_data=(X_test, y_test),
                    verbose=1)  # Set verbose=1 for detailed training logs
```

```python
# Evaluate the model on test data (optional but recommended)
train_loss, ann_train_accuracy = model.evaluate(X_train, y_train, verbose=0)

print(f"Train accuracy: {ann_train_accuracy:.4f}")
print(f"Train loss: {train_loss:.4f}")

test_loss, ann_test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test accuracy: {ann_test_accuracy:.4f}")
print(f"Test loss: {test_loss:.4f}")
```
